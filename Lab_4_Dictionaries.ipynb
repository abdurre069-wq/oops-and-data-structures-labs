<div style="background:#2563eb;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟦 Task 1 — Create an Energy Reading</h3></div>

> **🎯 Goal:** Create one time-stamped operating record and access values by key.

Named keys make records easier to understand than numeric positions.

```python
reading = {
    "time": "12:00",
    "load_kw": 2.4,
    "pv_kw": 1.9,
    "battery_soc_percent": 76,
    "grid_import_kw": 0.5
}
```

```python
print("Load:", reading["load_kw"], "kW")
print("Battery SOC:", reading["battery_soc_percent"], "%")
```

<div style="background:#059669;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟩 Task 2 — Add, Update, and Remove Fields</h3></div>

> **🎯 Goal:** Modify a record after receiving new measurements.

Assigning a new key adds a field; assigning an existing key updates it. `pop()` removes a field and returns its value.

```python
reading["electricity_price_pkr_per_kwh"] = 45.0
reading["battery_soc_percent"] = 74
removed_grid_value = reading.pop("grid_import_kw")
```

```python

print("Removed value:", removed_grid_value)
print("Number of fields:", len(reading))
print("Keys:", reading.keys())
```

<div style="background:#d1fae5;padding:12px 16px;border-left:5px solid #059669;border-radius:6px;"><b>💡 Result & Reflection</b><br>A new key-value pair is added when a new key is assigned. <code>pop()</code> removes the requested key and returns its stored value.<br><br><b>Questions:</b> What happens when a new key is assigned? What does <code>pop()</code> return?</div>

<div style="background:#d97706;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟨 Task 3 — Safe Access with get()</h3></div>

> **🎯 Goal:** Retrieve a field that may not be available.

Use `get()` with a default value when a dictionary field is optional.

```python
carbon_intensity = reading.get("carbon_intensity", "Not available")
print("Carbon intensity:", carbon_intensity)
```

<div style="background:#fef3c7;padding:12px 16px;border-left:5px solid #d97706;border-radius:6px;"><b>⚠️ Key Point</b><br>Square-bracket access raises a <code>KeyError</code> if the key is missing. <code>get()</code> can return a default value instead, making it useful for optional fields.<br><br><b>Question:</b> Why can <code>get()</code> be safer than square brackets?</div>

<div style="background:#7c3aed;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟪 Task 4 — Nested Microgrid Record</h3></div>

> **🎯 Goal:** Organize PV, battery, grid, and load information in nested dictionaries.

```python
microgrid = {
    "timestamp": "2026-09-30 14:00",
    "pv": {"power_kw": 3.2, "available": True},
    "battery": {"soc_percent": 68, "power_kw": -0.8},
    "grid": {"import_kw": 0.0, "export_kw": 0.6},
    "load": {"power_kw": 1.8, "critical": True}
}
```

```python

print("PV power:", microgrid["pv"]["power_kw"], "kW")
print("Battery SOC:", microgrid["battery"]["soc_percent"], "%")
```

```python
microgrid["battery"]["soc_percent"] = 66
microgrid["grid"]["export_kw"] = 0.8
```

```python

print("Updated battery SOC:", microgrid["battery"]["soc_percent"], "%")
print("Updated grid export:", microgrid["grid"]["export_kw"], "kW")
```

<div style="background:#ede9fe;padding:12px 16px;border-left:5px solid #7c3aed;border-radius:6px;"><b>🧠 Interpret the Result</b><br>The battery power is <code>-0.8 kW</code>. According to the lab's convention, negative battery power means the battery is charging.</div>

<div style="background:#db2777;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟥 Task 5 — Metadata and Time-Series Values</h3></div>

> **🎯 Goal:** Use a dictionary whose values include lists, then calculate net demand.

```python
day_data = {
    "date": "2026-09-30",
    "hour": [8, 9, 10, 11],
    "load_kw": [1.5, 1.8, 2.1, 2.0],
    "pv_kw": [0.4, 1.0, 1.8, 2.2]
}
```

```python
net_demand_kw = [
    load - pv
    for load, pv in zip(day_data["load_kw"], day_data["pv_kw"])
]
```

```python

day_data["net_demand_kw"] = net_demand_kw
print(day_data)
```

<div style="background:#fce7f3;padding:12px 16px;border-left:5px solid #db2777;border-radius:6px;"><b>📌 Observation</b><br>This structure keeps metadata and matching time-series values together. A negative net-demand value means PV generation is greater than load at that time.<br><br><b>Questions:</b> Why is this useful before pandas? What does negative net demand indicate?</div>

<div style="background:#0f766e;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟦 Task 6 — Update a Device Inventory</h3></div>

> **🎯 Goal:** Merge new maintenance information into an equipment record.

```python
inverter = {
    "device_id": "INV-01",
    "rated_power_kw": 5.0,
    "status": "operational"
}
```

```python
inverter.update({
    "last_service": "2026-09-15",
    "efficiency_percent": 96.5
})
```

```python

for key, value in inverter.items():
    print(key, ":", value)
```

<div style="background:#ccfbf1;padding:12px 16px;border-left:5px solid #0f766e;border-radius:6px;"><b>🔧 Key Point</b><br><code>update()</code> adds the supplied fields and overwrites values for keys that already exist.</div>

<div style="background:#0f766e;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🌞 Independent Lab-Report Task — Home Energy Management Record</h3></div>

> **Your challenge:** Build a home energy record with timestamp, weather, electricity price, load, PV, and battery data. Use nested dictionaries. Calculate net grid power as load minus PV minus battery discharge. Convention: positive battery power means discharging; negative means charging.

```python
home_record = {
    "timestamp": "2026-09-30 18:00",
    "weather": {"temperature_c": 29, "condition": "clear"},
    "electricity_price_pkr_per_kwh": 45.0,
    "load": {"power_kw": 2.4},
    "pv": {"power_kw": 0.7},
    "battery": {"soc_percent": 68, "battery_power_kw": 0.5}
}
```

```python
net_grid_power_kw = (
    home_record["load"]["power_kw"]
    - home_record["pv"]["power_kw"]
    - home_record["battery"]["battery_power_kw"]
)
home_record["net_grid_power_kw"] = net_grid_power_kw
```

```python
print("Home Energy Management Summary")
print("Timestamp:", home_record["timestamp"])
print("Weather:", home_record["weather"]["condition"], home_record["weather"]["temperature_c"], "°C")
```

```python

print(f"Net grid power: {net_grid_power_kw:.2f} kW")
print("Grid status:", "Importing" if net_grid_power_kw > 0 else "Exporting" if net_grid_power_kw < 0 else "Balanced")
```
