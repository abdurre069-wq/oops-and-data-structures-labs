<p>Return an operating instruction based on battery state of charge (SOC).</p>
</div>

````python
def battery_action(soc_percent: float) -> str:
    """Return a simple charging instruction from battery SOC."""
    if not 0 <= soc_percent <= 100:
        return "Invalid SOC"
    if soc_percent < 20:
        return "Charging required"
    if soc_percent > 90:
        return "Battery full; no charging"
    return f"Normal charging range; SOC = {soc_percent:.1f}%"

````

<h3 style="color:#553c9a;margin:0;">Task 4: Series and parallel resistance</h2>
<p>Write separate functions for two circuit calculations.</p>
</div>

````python
def series_resistance(resistances: list[float]) -> float:
    """Return the equivalent series resistance in ohms."""
    return sum(resistances)
````

````python

def parallel_resistance(resistances: list[float]) -> float:
    """Return the equivalent parallel resistance in ohms."""
    if not resistances or any(r <= 0 for r in resistances):
        raise ValueError("Resistances must be positive and non-empty.")
    return 1 / sum(1 / r for r in resistances)
````

````python


values = [10.0, 20.0, 30.0]
print("Series:", series_resistance(values), "ohm")
print("Parallel:", parallel_resistance(values), "ohm")
````

<h3 style="color:#166534;margin:0;">Task 7: Functions for energy-management features</h3>
<p>Calculate net grid power and electricity cost using small reusable functions.</p>
</div>

````python
def net_grid_power(load_kw: float, pv_kw: float, battery_kw: float = 0.0) -> float:
    """Return grid power; positive means import, negative means export."""
    return load_kw - pv_kw - battery_kw
````

````python

def energy_cost(grid_power_kw: float, hours: float, tariff: float) -> float:
    """Return import cost; exported energy is assigned zero cost here."""
    imported_energy_kwh = max(grid_power_kw, 0.0) * hours
    return imported_energy_kwh * tariff
````

````python

net_kw = net_grid_power(load_kw=3.5, pv_kw=2.0, battery_kw=0.8)
cost = energy_cost(net_kw, hours=1.0, tariff=45.0)
print(f"Net grid power: {net_kw:.2f} kW")
print(f"Hourly grid cost: {cost:.2f} PKR")
````

````python
# Quick Practice — solutions

def calculate_power(voltage: float, current: float, power_factor: float = 1.0) -> float:
    """Return active power in watts."""
    return voltage * current * power_factor
````

````python
def average_load(*loads: float) -> float:
    """Return the mean load; reject an empty input."""
    if not loads:
        raise ValueError("At least one load is required.")
    return sum(loads) / len(loads)
````

````python
def forecast_error(actual: float, predicted: float) -> float:
    """Return forecast error: actual minus predicted."""
    return actual - predicted
````

<div style="background:#fff4d6;padding:10px;border-radius:8px;"><b>Important convention:</b> <code>battery_power_kw &gt; 0</code> means discharge; <code>battery_power_kw &lt; 0</code> means charge.</div>

````python
# Home Energy Management Toolkit — solution

def calculate_net_grid_power(load_kw: float, pv_kw: float,
                             battery_power_kw: float = 0.0) -> float:
    """Return grid power in kW; positive is import, negative is export.

    Positive battery power means discharge; negative means charging.
    """
    return load_kw - pv_kw - battery_power_kw
````

````python

def calculate_import_cost(grid_power_kw: float, hours: float,
                          tariff_pkr_per_kwh: float) -> float:
    """Return the cost of imported energy in PKR; export cost is zero."""
    imported_energy_kwh = max(grid_power_kw, 0.0) * hours
    return imported_energy_kwh * tariff_pkr_per_kwh
````

````python

def estimate_self_consumption(load_kw: float, pv_kw: float) -> float:
    """Estimate PV self-consumption percentage for simultaneous load and PV."""
    if pv_kw < 0 or load_kw < 0:
        raise ValueError("Load and PV power must be non-negative.")
    if pv_kw == 0:
        return 0.0
    self_consumed_pv_kw = min(load_kw, pv_kw)
    return (self_consumed_pv_kw / pv_kw) * 100
````

````python

def battery_instruction(soc_percent: float) -> str:
    """Return a battery operating instruction based on SOC percentage."""
    if not 0 <= soc_percent <= 100:
        return "Invalid SOC"
    if soc_percent < 20:
        return "Charging required"
    if soc_percent > 90:
        return "Battery full; no charging"
    return f"Normal operating range; SOC = {soc_percent:.1f}%"
````

````python


for number, record in enumerate(records, start=1):
    grid_kw = calculate_net_grid_power(
        record["load_kw"], record["pv_kw"], record["battery_power_kw"]
    )
    cost = calculate_import_cost(grid_kw, hours, tariff)
    self_consumption = estimate_self_consumption(
        record["load_kw"], record["pv_kw"]
    )
    instruction = battery_instruction(record["soc"])
````
