### 🔌 Why This Matters
Lists are useful for measurements that may change, such as hourly load or solar generation. Tuples are useful for fixed information, such as phase labels and equipment ratings.

| Structure | Best used for | Energy example |
|---|---|---|
| List | Ordered data that may change | Hourly load: `[1.2, 1.5, 1.8]` |
| Tuple | Ordered data intended to stay fixed | Phase labels: `("R", "Y", "B")` |
| Nested list | Rows of related values | `[hour, load, solar, SOC]` |

<div style="background:#059669;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟩 Task 2 — Update and Extend Measurements</h3></div>

> **🎯 Goal:** Correct one meter value and add new readings.

Lists are mutable, so you can replace an item, append one value, or extend a list with multiple values.

```python
hourly_load_kw = [1.2, 1.4, 1.8, 2.2]
```

```python
hourly_load_kw[1] = 1.5
```

```python
hourly_load_kw.append(2.0)
```

```python
hourly_load_kw.extend([1.7, 1.6])
```

```python

print("Updated readings:", hourly_load_kw)
print("Total measured energy for one-hour intervals:", sum(hourly_load_kw), "kWh")
```

<div style="background:#dbeafe;padding:12px 16px;border-left:5px solid #2563eb;border-radius:6px;">
    <b>💡 Result & Reflection</b><br>
    <code>append()</code> adds one item; <code>extend()</code> adds each item from another iterable. For one-hour intervals, summing kW readings gives the numerical energy in kWh.<br><br>
    <b>Question:</b> Why can values in a list be updated?
</div>

<div style="background:#d97706;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟨 Task 3 — Filter High-Load Periods</h3></div>

> **🎯 Goal:** Use a list comprehension to find readings greater than 2.0 kW.

A list comprehension creates a new list from items that satisfy a condition.

```python
hourly_load_kw = [1.2, 2.4, 1.8, 2.7, 2.0, 1.6]
```

```python
high_load_kw = [load for load in hourly_load_kw if load > 2.0]
```

```python
print("High-load readings:", high_load_kw)
```

<div style="background:#fef3c7;padding:12px 16px;border-left:5px solid #d97706;border-radius:6px;"><b>⚡ Observation</b><br>Only values strictly greater than 2.0 kW are included; a value equal to 2.0 is excluded.<br><br><b>Question:</b> How could identifying high-load periods help a demand-side management controller?</div>

<div style="background:#7c3aed;color:white;padding:14px 18px;border-radius:10px;"><h3 style="color:white;margin:0;">🟪 Task 4 — Nested Energy Records</h3></div>

> **🎯 Goal:** Represent each hour as `[hour, load, solar, battery SOC]`.

Nested lists store related fields in each row.

```python
energy_data = [
    [8, 1.6, 0.5, 82],
    [9, 1.8, 1.1, 78],
    [10, 2.0, 1.7, 75]
]
```

```python
print("Solar at 09:00:", energy_data[1][2], "kW")
```

```python
energy_data[2][3] = 73
energy_data.append([11, 2.1, 2.0, 71])
```

```python
print("Updated energy data:", energy_data)
```

<div style="background:#ede9fe;padding:12px 16px;border-left:5px solid #7c3aed;border-radius:6px;"><b>🧠 Interpret the Result</b><br><code>energy_data[2][3]</code> refers to the battery SOC in the third row (hour 10), using zero-based indexing.<br><br><b>Limitation:</b> Numeric positions can be hard to remember and may cause mistakes if the row structure changes.</div>

> **🎯 Goal:** Use a tuple for device ratings and demonstrate immutability.

```python
inverter_rating = ("INV-01", 5.0, 230, 0.95)
```

```python
print("Device ID:", inverter_rating[0])
print("Rated power:", inverter_rating[1], "kW")
```

<div style="background:#fce7f3;padding:12px 16px;border-left:5px solid #db2777;border-radius:6px;"><b>🔐 Observation</b><br>A tuple's items cannot be reassigned, making it suitable for fixed manufacturer-rated values.<br><br><b>Question:</b> Why is a tuple reasonable for equipment ratings?</div>

> **Your challenge:** Use separate lists for load power, PV power, and battery SOC. Calculate net demand (load minus PV), maximum load, total PV energy for one-hour intervals, and hours with positive net demand. Store fixed site information in a tuple.

```python
site_info = ("UET Microgrid", 230, 50)  # site name, voltage (V), frequency (Hz)
```

```python
load_kw = [2.0, 2.2, 2.5, 2.4, 2.1, 1.8]
pv_kw = [0.2, 0.8, 1.5, 2.0, 1.2, 0.3]
battery_soc = [80, 78, 75, 74, 76, 79]
```

```python
net_demand_kw = [load - pv for load, pv in zip(load_kw, pv_kw)]
max_load_kw = max(load_kw)
total_pv_energy_kwh = sum(pv_kw)  # one-hour intervals
positive_net_demand_hours = [hour for hour, demand in enumerate(net_demand_kw) if demand > 0]
```

```python
print("Site information:", site_info)
print("Net demand by hour:", net_demand_kw)
print(f"Maximum load: {max_load_kw:.2f} kW")
```

```python
print(f"Total PV energy: {total_pv_energy_kwh:.2f} kWh")
print("Hours with positive net demand (0-based):", positive_net_demand_hours)
```
