<div style="background:#e0f2fe; color:#0c4a6e; padding:10px 14px; border-left:5px solid #0284c7; border-radius:6px;">**🎯 Goal:** Convert sensor readings received as text into numbers, then calculate electrical power.

**Why it matters:** Sensor data may arrive as text. Python needs numeric values to perform calculations, so we'll convert the readings before multiplying.</div>

```python
voltage_text = "230.5"
current_text = "4.2"
```

```python
voltage = float(voltage_text)
current = float(current_text)
power = voltage * current
```

```python
print(f"Voltage: {voltage:.1f} V")
print(f"Current: {current:.2f} A")
print(f"Power: {power:.2f} W")
```

### 💡 Result & Reflection

**Observation:** Type conversion allows numeric calculations on readings received as strings.

**Question:** What error occurs if the text is not converted before multiplication?

**Answer:** Multiplying two strings does not perform numeric multiplication; Python raises a `TypeError` for string-by-string multiplication.

<div style="background-color:#7c3aed; color:white; padding:14px 18px; border-radius:10px; margin:10px 0 16px 0;"><h2 style="color:white; margin:0;">🟪 Complex Impedance of a Series RLC Circuit</h2></div>

<div style="background:#e0f2fe; color:#0c4a6e; padding:10px 14px; border-left:5px solid #0284c7; border-radius:6px;">**🎯 Goal:** Calculate resistance, net reactance, impedance magnitude, phase, and current using complex numbers.

```python
import math
import cmath

```

```python
resistance = 20.0       # ohm
inductance = 0.08       # H
capacitance = 100e-6     # F
frequency = 50.0        # Hz
source_voltage = 230.0  # V (RMS)

```

```python
omega = 2 * math.pi * frequency
z_r = complex(resistance, 0)
z_l = complex(0, omega * inductance)
z_c = complex(0, -1 / (omega * capacitance))
z_total = z_r + z_l + z_c
current = source_voltage / z_total

```

```python
print(f"Total impedance: {z_total:.2f} ohm")
print(f"Resistance: {z_total.real:.2f} ohm")
print(f"Net reactance: {z_total.imag:.2f} ohm")
print(f"|Z|: {abs(z_total):.2f} ohm")
print(f"Phase: {math.degrees(cmath.phase(z_total)):.2f} degree")
print(f"Current magnitude: {abs(current):.2f} A")
```

### 🧠 Interpret the Result

**Observation:** The sign of the imaginary part indicates whether the net reactance is inductive or capacitive. A positive imaginary part means inductive; a negative imaginary part means capacitive.

**Question:** Is this circuit predominantly inductive or capacitive?

**Answer:** Compare the sign of `z_total.imag` with zero after running the code.
